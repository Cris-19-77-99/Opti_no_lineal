```python
import numpy as np
import sympy as sp
import math

def f(v):
  return (v[0]-4)**(4)+(v[1]-3)**(2)+4*(v[2]+5)**(4)
def df(v):
  return np.array([[4*((v[0]-4)**3)],[2*(v[1]-3)],[16*((v[2]+5)**3)]])
def mdsec(y0,y1,df_f):
  y2= y1-((y1-y0)/(df_f(y1)-df_f(y0)))*df_f(y1)

  if df_f(y2) < 10**(-5):
    return (y2)

  else:
    return(mdsec(y1,y2,df_f))


def fc(f,df,x0,d0,c):
  c=c+1
  if math.sqrt((df(x0)[0])**2+(df(x0)[1])**2+(df(x0)[2])**2)< 10**(-5):
    return (x0,c)
  else:
    print(x0)
    def df_a(z):
      x = sp.Symbol('x')
      fun=f(x0+np.transpose(x*d0)[0])
      f_prime=fun.diff(x)
      f_prime = sp.lambdify(x, f_prime)
      return(f_prime(z))
    print(df_a(2))
    a= mdsec(-2,2,df_a)
    x1=(x0+np.transpose(a*d0))[0]
    if math.sqrt((df(x1)[0])**2+((df(x1)[1])**2+(df(x1)[2])**2))< 10**(-5):
      return (x1,c)
    else:
      b= max(0,((np.transpose(df(x1))@(df(x1)-df(x0)))/(np.transpose(d0)@(df(x1)-df(x0))))[0])
      d1= -df(x1)+b*d0
      return(fc(f,df,x1,d1,c))
fc(f,df,np.array([4,2,1]),-df(np.array([4,2,1])),0)
```
